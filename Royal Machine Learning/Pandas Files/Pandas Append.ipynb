# Pandas Join() or Append() --> Replacement of Concat() :

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'A':[1,2,3,4],
                    'B':[10,20,30,40]})
df2 =pd.DataFrame({'A':[5,6,7,8],
                   'B':[50,60,70,80]})
display(df1,df2)
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], ignore_index= True)
```

```python
pd.concat([df1, df2], ignore_index=True, sort=False)
```
