```python
classes = ["Python", "R", "Machine Learning", "Artificial Intelligence", "Data Science"]
cls1_studs = [45,15,35,25,30]
explode =[0.03,0,0.1,0,0]

plt.pie(cls1_studs, labels = classes,explode= explode)
plt.show()

```

```python
classes = ["Python", "R", "Machine Learning", "Artificial Intelligence", "Data Science"]
cls1_studs = [45,15,35,25,30]
explode =[0.03,0,0.1,0,0]
colors = ["c", "b", "r", "y", "g"]

plt.pie(cls1_studs, labels = classes,explode= explode, colors =colors, autopct = "%0.2f%%", shadow= True)
plt.show()

```

```python
classes = ["Python", "R", "Machine Learning", "Artificial Intelligence", "Data Science"]
cls1_studs = [45,15,35,25,30]
explode =[0.03,0,0.1,0,0]
colors = ["c", "b", "r", "y", "g"]

plt.pie(cls1_studs, labels = classes,explode= explode, colors =colors, autopct = "%0.2f%%", shadow= True,
        radius = 1.4, startangle =90)
plt.show()

```

```python
classes = ["Python", "R", "Machine Learning", "Artificial Intelligence", "Data Science"]
cls1_studs = [45,15,35,25,30]
explode =[0.03,0,0.1,0,0]
colors = ["c", "b", "r", "y", "g"]
textprops = {"fontsize": 15}

plt.pie(cls1_studs, labels = classes,explode= explode, colors =colors, autopct = "%0.2f%%", shadow= True,
        radius = 1.4, startangle = 270, textprops=textprops)
plt.show()

```

```python
plt.Figure(figsize=(16,8))
wedgeprops ={"linewidth": 4, "width":3, "edgecolor":"k"}
plt.pie(
        cls1_studs,
        labels= classes,
        explode=explode,
        colors=colors,
        autopct="%0.2f%%",
        pctdistance= 0.6,
        shadow=True,
        labeldistance= 1.2,
        startangle= 270,
        radius= 1.8,
        counterclock = True,
        wedgeprops = wedgeprops,
        textprops = textprops,
        center =(1,5),
        frame=True,
        rotatelabels=True
)
plt.show()
```

```python
cls1_studs = [45,15,35,25,30]
explode =[0.03,0,0.1,0,0]
colors = ["c", "b", "r", "y", "g"]
textprops = {"fontsize": 15}

plt.pie(cls1_studs, labels = classes,explode= explode, colors =colors, autopct = "%0.2f%%", shadow= True,
        radius = 1.4, startangle = 270, textprops=textprops)
plt.legend()
plt.show()
```
