 # Ploting Bar Chart :

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

```python
classes= ["Python", "R", "AI", "ML", "DS"]
cls1_studs =[30,10,20,25,10]
cls2_studs = [40,5,20,20,10]
cls3_studs = [35,5,30,15,15]
```

```python
plt.bar(classes, cls1_studs)
```

```python
plt.barh(classes, cls1_studs)
```

```python
plt.bar(classes, cls1_studs, width=0.2, align="edge")
```

```python
plt.bar(classes, cls1_studs, width=0.2, align="edge", color ='y', edgecolor="g", linewidth=5, alpha=0.5)
```

```python
plt.bar(classes, cls1_studs, width=0.2, align="edge", color ='y', edgecolor="g", linewidth=5, alpha=0.5,
        linestyle="--", label="Class 1 Students")
```

```python
plt.bar(classes, cls1_studs, width=0.2, align="edge", color ='y', edgecolor="g", linewidth=5, alpha=0.5,
        linestyle="--", label="Class 1 Students", visible=False)
```

```python
style.use("ggplot")
plt.figure(figsize=(16,9))
plt.bar(classes, cls1_studs, width=0.6, align="edge", color ='k', edgecolor="g", linewidth=5, alpha=0.5,
        linestyle="--", label="Class 1 Students")
```

```python
plt.figure(figsize=(16,9))
plt.barh(classes, cls1_studs, align="edge", color ="k", edgecolor="g", linewidth=5, alpha=0.5,
        linestyle="--", label="Class 1 Students")
```

```python
plt.figure(figsize=(16,9))
plt.bar(classes, cls1_studs, width=0.6, align="edge", color ='k', edgecolor="g", linewidth=5, alpha=0.5,
        linestyle="--", label="Class 1 Students")

plt.title("Bar Chart of PR Class", fontsize=18)
plt.xlabel("Classes", fontsize=15)
plt.ylabel("No. Of Studs", fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))
plt.bar(classes, cls1_studs, width=0.3, color ='b', label="Class 1 Students")

plt.bar(classes, cls2_studs, width=0.3, color ='g',label="Class 2 Students")

plt.title("Bar Chart of PR Class", fontsize=18)
plt.xlabel("Classes", fontsize=15)
plt.ylabel("No. Of Studs", fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))

classes_index =np.arange(len(classes))
width =0.3

plt.bar(classes_index, cls1_studs, width, color ='b', label="Class 1 Students")

plt.bar(classes_index + width, cls2_studs, width, color ='g',label="Class 2 Students")

plt.xticks(classes_index + width, classes, rotation =20)
plt.title("Bar Chart of PR Class", fontsize=18)
plt.xlabel("Classes", fontsize=15)
plt.ylabel("No. Of Studs", fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))

classes_index =np.arange(len(classes))
width =0.3

plt.bar(classes_index, cls1_studs, width, color ='b', label="Class 1 Students")

plt.bar(classes_index + width, cls2_studs, width, color ='g',label="Class 2 Students")

plt.bar(classes_index + width, cls3_studs, width, color ='m',label="Class 2 Students")


plt.xticks(classes_index + width, classes, rotation =20)
plt.title("Bar Chart of PR Class", fontsize=18)
plt.xlabel("Classes", fontsize=15)
plt.ylabel("No. Of Studs", fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))

classes_index =np.arange(len(classes))
width =0.3

plt.bar(classes_index, cls1_studs, width, color ='b', label="Class 1 Students")

plt.bar(classes_index + width, cls2_studs, width, color ='g',label="Class 2 Students")

plt.bar(classes_index + width + width, cls3_studs, width, color ='m',label="Class 2 Students")


plt.xticks(classes_index + width, classes, rotation =20)
plt.title("Bar Chart of PR Class", fontsize=18)
plt.xlabel("Classes", fontsize=15)
plt.ylabel("No. Of Studs", fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))

classes_index =np.arange(len(classes))
width =0.3

plt.barh(classes_index, cls1_studs, width, color ='b', label="Class 1 Students")

plt.barh(classes_index + width, cls2_studs, width, color ='g',label="Class 2 Students")

plt.barh(classes_index + width + width, cls3_studs, width, color ='m',label="Class 3 Students")


plt.yticks(classes_index + width, classes, rotation =20)
plt.title("Bar Chart of PR Class", fontsize=18)
plt.ylabel("Classes", fontsize=15)
plt.xlabel("No. Of Studs", fontsize=15)
plt.legend()
plt.show()
```
