###  Matrix Multiplication 
Two main ways of performing multiplication in neural networks and deep learning :
1.Elements wise multipliaction 
2.Matrix multiplication (dot product)

more information on multiply matrics -https://www.mathsisfun.com/algebra/matrix-multiplying.html

There are  two main that perform matrix multiplication needs to satisfy:
1. the **inner dimensions ** must match:
* '(3,2) @ ( 3,2)' wont't work
*'(2,3) @ (3,2)' will work
* '(3,2) @  (2,3)' will work

2. The resulting matrix has the shape of the ** outer dimensions ** :
*  '(2,3 ) @  (3,2)' -> (2,2)
* '(3,2) @ (2,3)' -> (3,3)



```python
torch.matmul(torch.rand(10, 10), torch.rand(10, 10)).shape
```

### One of the common error in Deep Learning shape errors :


```python
# Shape for matrix multipliaction 
tensor_A = torch.tensor([[1,2],
                        [3,4],
                        [5,6]])

tensor_B =torch.tensor([[7,10],
                        [8,11],
                        [9,12]])

# torch.mm(tensor_A, tensor_B) # torch.mm is the same as torch.matmul(it's an alis for)
torch.matmul(tensor_A, tensor_B)
```

```python
tensor_A.shape, tensor_B.shape
```

To fix our tensor shape issue, we can manipulates the one of our tensor using a **transpose**.

A **transpose** switches  the axes or dimensions of a given tensor.

```python
tensor_B.T, 
```

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
```

```python
# The Matix multiplication operation works when tensor_B is transposed:
print(f"Original shape : tensor_A = {tensor_A.shape}, tensor_B ={tensor_B.shape}")
print(f"New shapes : tensor_A = {tensor_A.shape} (same shape as above), tensor_B.T = {tensor_B.T.shape}")
print(f"Multiplying : {tensor_A.shape } @ {tensor_B.T.shape} <- innetr dimensions must match")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}") 

```

## Reshaping, Stacking, Squeezing and unsqueezing tensors

* Reshaping - reshape an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory  as the original tensor
* Stacking - combine multiple tensors on top each other (vstack) or side by side (hstack)
* Squeez - remove all '1' dimension from a tensor
* Unsqueez - add a '1' dimension to a target tensor
* Permute - Return a view of the input with dimension permuted (swapped) in a certain way

```python
# Lets create a tensor 
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimensions
x_reshaped =x.reshape(1,9)
x_reshaped, x_reshaped.shape
```

```python
# Add an extra dimensions
x_reshaped =x.reshape(9, 1)
x_reshaped, x_reshaped.shape
```

```python
# change  the view 
z = x.view(1, 9)
z, z.shape
```

```python
# Changing z changes x (becaues a view of a tensor shares the same memory as the original input)
z[:, 0] =5
z, x
```

```python
# Stack tensors on top of each other
x_stacked  = torch.stack([x, x, x, x], dim=0)
x_stacked
```

```python
# Stack tensors on top of each other
x_stacked  = torch.stack([x, x, x, x], dim=1)
x_stacked
```
