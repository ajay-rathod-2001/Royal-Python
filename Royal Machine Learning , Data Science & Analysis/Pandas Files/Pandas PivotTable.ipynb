# Pandas Pivot_Table() :

```python
import pandas as pd
```

```python
df = pd.read_csv("C:\\Users\\ajayr\\Downloads\\Diabetes Missing Data.csv")
df
```

```python
df.pivot_table(index='Pregnant')
```

```python
df.pivot_table(index='Pregnant', columns='Skin_Fold')
```

```python
df.pivot_table(index='Pregnant', columns='Skin_Fold', aggfunc= 'count')
```

```python
df.pivot_table(index='Pregnant', columns='Skin_Fold', aggfunc= 'max')
```

```python
df.pivot_table(index='Pregnant', columns='Skin_Fold', aggfunc= 'sum')
```

```python
df.pivot_table(index='Pregnant', columns='Skin_Fold', fill_value=0)
```

```python
df.pivot_table(index='Pregnant', columns='Skin_Fold', fill_value=0, margins= True)
```

```python
df.pivot_table(index='Pregnant', columns='Skin_Fold', aggfunc='sum',fill_value=0, margins= True)
```
