Python NumPy Array ConCatination And Split :


```python
import numpy as np
```

```python
arr1 = np.arange(1,17).reshape(4,4)
print(arr1)
```

```python
arr2 = np.arange(17,33).reshape(4,4)
print(arr2)
```

```python
list1 = [2,4,6,7]
list2 = [1,8,3,5]
```

```python
list1 + list2
```

```python
arr1 + arr2
```

```python
np.concatenate((arr1, arr2))
```

```python
 np.concatenate((arr1,arr2), axis =1)
```

```python
np.vstack((arr1,arr2))
```

```python
np.hstack((arr1,arr2))
```

```python
np.hstack((arr1,arr2,arr1))
```
