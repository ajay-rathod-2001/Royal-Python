## How to Draw Seaborn HetMap :

```python
# Import Libraries 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
arr_2d = np.linspace(1,5,12).reshape(4,3)
arr_2d
```

```python
sns.heatmap(arr_2d)
```

```python
gw_df = pd.read_csv("C:\\Users\\ajayr\\Downloads\\Who_is_responsible_for_global_warming.csv")
gw_df
```

```python
gw_df = gw_df.drop(columns=["Country Code", "Indicator Name", "Indicator Code"]).set_index("Country Name")
gw_df.head(6)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(gw_df)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(gw_df, vmin =0, vmax=21)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(gw_df, cmap="coolwarm")
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(gw_df, cmap="coolwarm", annot = True)
```

```python
arr_2d
```

```python
# fmp parameter - add text on heatmap cell
 
annot_arr = np.array([['a00','a01','a02'],
                      ['a10','a11','a12'],
                      ['a20','a21','a22'],
                      ['a30','a31','a32']], dtype = str)
 
annot_arr
```

```python
sns.heatmap(arr_2d, annot = annot_arr, fmt ="s" )
```

```python
plt.figure(figsize =(16,9))
annot_kws ={"fontsize":10,
             "fontstyle": "italic",
             "color": "k",
             "alpha":0.9,
             "rotation": "vertical",
             "verticalalignment": "center",
             "backgroundcolor": "w"}
sns.heatmap(gw_df, cmap = "coolwarm", annot = True, annot_kws = annot_kws)
```

```python
plt.figure(figsize = (16,9))
sns.heatmap(gw_df ,  cmap ="coolwarm", annot = True, linewidth = 4)
```

```python
plt.figure(figsize = (16,9))
sns.heatmap(gw_df ,  cmap ="coolwarm", annot = True, linewidth = 4, linecolor="k")
```

```python
plt.figure(figsize =(16,9))
sns.heatmap(gw_df, cbar = False)
```

```python
plt.figure(figsize =(16,9))
sns.heatmap(gw_df, cbar = False, xticklabels = False, yticklabels = False)
```

```python
plt.figure(figsize =(16,9))
cbar_kws ={"orientation": "horizontal",
           "shrink": 1,
           "extend": "min",
           "extendfrac":0.1,
           "ticks":np.arange(0,22),
           "drawedges": True}
sns.heatmap(gw_df, cbar_kws = cbar_kws )
```

```python
plt.figure(figsize =(16,9))
Country_Code = ["USA", "BGR", "IND", "CHN", "RUS", "AUS", "FRA", "DEU", "CAN", "BRA", "ARG", "PAK", "NPL", "BGD", "JPN"]
sns.heatmap(gw_df, xticklabels = np.arange(0,15), yticklabels = Country_Code)
```

```python
plt.figure(figsize=(16,9))
ax = sns.heatmap(gw_df)
ax.set(title = "Heatmap of 'Who is Resposible for Global Warming'",
       xlabel= "CO2 Emissions (metric tons per capita ) per Year",
       ylabel = "Country Name")
sns.set(font_scale= 2)
```
