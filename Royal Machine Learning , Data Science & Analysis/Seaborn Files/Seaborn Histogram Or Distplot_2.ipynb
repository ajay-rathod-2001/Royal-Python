## Hoe to Draw Seaborn Histogram / Seaborn Distplot or Displot (kdeplot) :

```python
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
tips_df = sns.load_dataset("tips")
tips_df
```

```python
sns.displot(tips_df["size"], kde=True) # updated New -->displot or histplot for new version of 'seaborn', flipher-->kdeplot 
# sns.distplot(tips_df["size"])
```

```python
plt.figure(figsize=(12,6))
sns.set()

sns.displot(tips_df["total_bill"], color='#DC143c', edgecolor = "#00ff44d9", kde=True,
            linewidth =2, linestyle = '--', alpha =0.9)
# sns.distplot(tips_df["total_bill"], hist_kws = {'color ': '#DC143c'})
plt.show()
```

```python
plt.figure(figsize =(13,7))
sns.set()

sns.distplot(
    tips_df["total_bill"],
    hist_kws={
        'color': '#DC143C',        # fill color for bars
        'edgecolor': '#00ff44d9',  # outline color
        'linewidth': 1,
        'linestyle': '--',
        'alpha': 0.9
    },
    kde_kws = {
         'color': "#0AB2EAE2",        # fill color for bars
        'linewidth': 2,
        'linestyle': '--',
        'alpha': 0.9
        
    },
 
    
    rug = True,
    rug_kws = {
        'color': "#0508D6",        # fill color for bars
        'edgecolor': '#00ff44d9',  # outline color
        'linewidth': 5,
        'linestyle': '--',
        'alpha': 0.9
        
    }
)

plt.show()
```

```python
plt.figure(figsize =(13,7))
sns.distplot(tips_df["total_bill"], bins =9, label = "total_bill")
sns.distplot(tips_df["tip"], bins =9, label = "tip")
sns.distplot(tips_df["size"], bins =9, label = "size")

plt.legend()


```

```python
fig, axes = plt.subplots(1, 3, figsize=(16,9))
sns.set(style="whitegrid")


sns.histplot(tips_df["total_bill"], bins=9, kde=True, ax=axes[0], color="#1f77b4", )
axes[0].set_title("Total Bill Distribution")


sns.histplot(tips_df["tip"], bins=9, kde=True, ax=axes[1], color="#ff7f0e", label = "Tip")
axes[1].set_title("Tip Distribution")

sns.histplot(tips_df["size"], bins=9, kde=True, ax=axes[2], color="#2ca02c", label = "Size")
axes[2].set_title("Size Distribution")
#plt.tight_layout()
plt.legend()
plt.show()

```

```python
fig, axes = plt.subplots(1, 3, figsize=(16,5))
sns.set(style="whitegrid")

# Total Bill
sns.histplot(
    tips_df["total_bill"],
    bins=9,
    kde=True,
    ax=axes[0],
    color="#1f77b4",
    label="Total Bill"
)
axes[0].set_title("Total Bill Distribution")
axes[0].legend()

# Tip
sns.histplot(
    tips_df["tip"],
    bins=9,
    kde=True,
    ax=axes[1],
    color="#ff7f0e",
    label="Tip"
)
axes[1].set_title("Tip Distribution")
axes[1].legend()

# Size
sns.histplot(
    tips_df["size"],
    bins=9,
    kde=True,
    ax=axes[2],
    color="#2ca02c",
    label="Size"
)
axes[2].set_title("Size Distribution")
axes[2].legend()

plt.tight_layout()
plt.show()
```
