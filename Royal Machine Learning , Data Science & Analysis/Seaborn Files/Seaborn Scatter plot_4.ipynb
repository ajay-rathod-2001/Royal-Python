# How tob Draw Seaborn Scatter Plot :  

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titatic_df = sns.load_dataset("titanic")
titatic_df
```

```python
sns.scatterplot(x = "age", y = "fare", data = titatic_df)
```

```python
sns.scatterplot(x = "age", y = "fare", data = titatic_df, hue = "sex")
```

```python
sns.scatterplot(x = "age", y = "fare", data = titatic_df, hue = "sex", style = "who")
```

```python
sns.scatterplot(x = "age", y = "fare", data = titatic_df, hue = "sex", style = "who",
                size = "who")
```

```python
sns.scatterplot(x = "age", y = "fare", data = titatic_df, hue = "sex", style = "who",
                size = "who", sizes =(100,400))
```

```python
plt.figure(figsize =(16,9))
sns.scatterplot(x = "age", y = "fare", data = titatic_df, hue = "sex", style = "who",
                size = "who", sizes =(100,400))
```

```python
plt.figure(figsize =(16,9))
sns.scatterplot(x = "who", y = "fare", data = titatic_df, hue = "alive", style = "alive",
                size = "who", sizes =(100,400))
```

```python
plt.figure(figsize =(16,9))
sns.scatterplot(x = "who", y = "fare", data = titatic_df, hue = "alive", style = "alive",
                size = "who", sizes =(100,400),palette = "inferno_r", alpha =.7)
```

```python
plt.figure(figsize =(35,17))
sns.scatterplot(x = "age", y  = "fare", data = titatic_df)
sns.lineplot( x = "age", y = "fare", data = titatic_df)
sns.barplot(x  = "age", y = "fare",  data =titatic_df)
plt.show()
```
