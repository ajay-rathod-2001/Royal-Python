```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df_google_play_store_apps =pd.read_csv("C:\\Users\\ajayr\\Downloads\\archive (1)\\googleplaystore.csv")
df_google_play_store_apps.shape
```

```python
df_google_play_store_apps =pd.read_csv("C:\\Users\\ajayr\\Downloads\\archive (1)\\googleplaystore.csv", nrows =1000)
df_google_play_store_apps.shape
```

```python
df_google_play_store_apps
```

```python
x = df_google_play_store_apps["Rating"]
y = df_google_play_store_apps["Reviews"]
```

```python
plt.scatter(x,y)
```

```python
plt.scatter(x,y)
plt.title("Google Play Store Apps Scatter Plot")
plt.xlabel("Rating")
plt.ylabel("Reviews")
```

```python
plt.figure(figsize=(16,9))
plt.scatter(x,y, c = "r", marker = "*", s = 100, alpha=0.5, linewidth = 7)
plt.title("Google Play Store Apps Scatter Plot")
plt.xlabel("Rating")
plt.ylabel("Reviews")
```

```python
plt.figure(figsize=(16,9))
plt.scatter(x,y, c = "r", marker = "*", s = 100, alpha=0.5, linewidth = 7, 
            edgecolors= "g" )
plt.title("Google Play Store Apps Scatter Plot")
plt.xlabel("Rating")
plt.ylabel("Reviews")
```

```python
plt.figure(figsize=(16,9))
plt.scatter(x,y, c = "r", marker = "*", s = 100, alpha=0.5, linewidth = 7, 
            edgecolors= "g" )
plt.scatter(x,df_google_play_store_apps["Installs"], c = "y", marker = "o", s = 100, alpha=0.5, linewidth = 7, 
            edgecolors= "c" )

plt.title("Google Play Store Apps Scatter Plot")
plt.xlabel("Rating")
plt.ylabel("Reviews & Installs")
```
