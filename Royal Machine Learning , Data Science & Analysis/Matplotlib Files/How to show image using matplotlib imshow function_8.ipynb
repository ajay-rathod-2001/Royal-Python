## How to show Image Using Matplotlib imshow Functuin() :

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
img = mpimg.imread("pie_chart.png")
```

```python
img
```

```python
type(img)
```

```python
img.ndim
```

```python
plt.figure(figsize=(16,9))
plt.axis("off")
plt.imshow(img)
plt.show()
```

```python
plt.figure(figsize=(12,7))
plt.axis("off")
plt.imshow(img)
plt.colorbar()
plt.show()
```

```python
single_channel = img[:,:,1]
plt.figure(figsize=(12,7))
plt.axis("off")
plt.imshow( single_channel,cmap="hot")
plt.colorbar()
plt.show()
```
