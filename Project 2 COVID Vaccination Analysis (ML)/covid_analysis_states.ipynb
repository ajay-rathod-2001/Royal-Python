## Fully Vaccinated


```python
fully_vaccinated =df.Fully_Vaccinated.tolist()
fully_vaccinated
```

```python
len_fully_vaccinated = len(fully_vaccinated)
len_fully_vaccinated
```

Now, It seems out clearly that length of people who are fully vaccinated (Actually which falls into that category; whether actually fully vaccinated or not), is 1000(nearby).Now, ur task is to find actuall list of people from the data frame which have fully vaccinated="yes".

# method 1:

```python
fully_vaccinated_yes =[]
for y in fully_vaccinated:
    if y=='Yes':
        fully_vaccinated_yes.append(y)
fully_vaccinated_yes
```

```python
len_yes=len(fully_vaccinated_yes)
len_yes
```

Now we have to find that how the len_yes ,i.e, full vaccination is how much %age of total vaccinated program

```python
confirm_full_vaccinated_percent =(100*len_yes)/len_fully_vaccinated
confirm_full_vaccinated_percent
```

```python
print("Total over all Vaccination in states = ",round(confirm_full_vaccinated_percent))
```

Now States wise Analysis :

```python
sts_yes=df.loc[df["State/U.T"]=="Maharashtra"]
sts_yes
```

```python
st_confirm_vaccination=sts_yes.loc[sts_yes["Fully_Vaccinated"]=="Yes"]
st_confirm_vaccination
```

```python
st_yes=st_confirm_vaccination["Fully_Vaccinated"].tolist()
st_yes
```

```python
l_y =len(st_yes)
l_y
```

```python
st_prcent_confirm =(l_y * 100/len_yes)
st_prcent_confirm 
```

```python
msg = "The %age of 2nd Vaccination confirm for Maharashtra is = "+str(round(st_prcent_confirm, 3))+"%."
print(msg)
```

```python
each_state_case = full_confirm_yes[full_confirm_yes["Fully_Vaccinated"]=="Yes"]
each_state_case
```

```python
state_wise = each_state_case.groupby("State/U.T").count()
state_wise
```

```python
st_full_yes = state_wise[["Fully_Vaccinated"]]
st_full_yes
```

```python
import matplotlib.pyplot as plt
import seaborn as sns, matplotlib
```

```python
matplotlib.rcParams["font.size"]=14
matplotlib.rcParams["figure.figsize"]=(9, 5)
matplotlib.rcParams["figure.facecolor"]="#00000000"
```

```python
%matplotlib inline
plt.show()
```

```python
plt.figure(figsize=(12,8))
st_full_yes.plot(kind="bar",
                 xlabel="State/U.T",
                 ylabel="Fully_Vaccinated",
                 color="red", width=.27,
                 title="State Wise Full Vaccination")
```

```python
plt.plot(st_full_yes["Fully_Vaccinated"])
plt.xticks(rotation=90);
```

```python
plt.title("Full Vaccination Status of India")
sns.heatmap(st_full_yes,cmap="Blues",annot=True,fmt="d");
```
