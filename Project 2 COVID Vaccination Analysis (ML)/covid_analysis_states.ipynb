Now, It seems out clearly that length of people who are fully vaccinated (Actually which falls into that category; whether actually fully vaccinated or not), is 1000(nearby).Now, ur task is to find actuall list of people from the data frame which have fully vaccinated="yes".

# method 1:

```python
fully_vaccinated_yes =[]
for y in fully_vaccinated:
    if y=='Yes':
        fully_vaccinated_yes.append(y)
fully_vaccinated_yes
```

```python
len_yes=len(fully_vaccinated_yes)
len_yes
```

Now we have to find that how the len_yes ,i.e, full vaccination is how much %age of total vaccinated program

```python
confirm_full_vaccinated_percent =(100*len_yes)/len_fully_vaccinated
confirm_full_vaccinated_percent
```

```python
print("Total over all Vaccination in states = ",round(confirm_full_vaccinated_percent))
```
